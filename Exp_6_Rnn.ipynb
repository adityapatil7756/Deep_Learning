```python
import tensorflow as tf
import keras
from keras.datasets import imdb
from keras.preprocessing.sequence import pad_sequences
from keras import layers, models
import matplotlib.pyplot as plt

vocab_size = 10000
max_len = 200

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=vocab_size)

x_train = pad_sequences(x_train, maxlen=max_len)
x_test = pad_sequences(x_test, maxlen=max_len)

rnn_model = models.Sequential()

rnn_model.add(layers.Embedding(vocab_size, 128))
rnn_model.add(layers.SimpleRNN(32))
rnn_model.add(layers.Dense(32, activation='relu'))
rnn_model.add(layers.Dense(1, activation='sigmoid'))

rnn_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

history = rnn_model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=64,
    validation_data=(x_test, y_test)
)

loss, acc = rnn_model.evaluate(x_test, y_test)
print("RNN Accuracy:", acc)

plt.plot(history.history['val_accuracy'], label='RNN Val Accuracy')
plt.plot(history.history['val_loss'], label='RNN Val Loss')
plt.legend()
plt.show()
```

```python
import tensorflow as tf
import keras
from keras.datasets import imdb
from keras.preprocessing.sequence import pad_sequences
from keras import layers, models
import matplotlib.pyplot as plt
vocab_size = 10000
max_len = 200

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=vocab_size)

x_train = pad_sequences(x_train, maxlen=max_len)
x_test = pad_sequences(x_test, maxlen=max_len)

gru_model = models.Sequential()

gru_model.add(layers.Embedding(vocab_size, 128))
gru_model.add(layers.GRU(64, dropout=0.3, recurrent_dropout=0.3))
gru_model.add(layers.Dense(32, activation='relu'))
gru_model.add(layers.Dense(1, activation='sigmoid'))

gru_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

history = gru_model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=64,
    validation_data=(x_test, y_test)
)

loss, acc = gru_model.evaluate(x_test, y_test)
print("GRU Accuracy:", acc)

plt.plot(history.history['val_accuracy'], label='GRU Val Accuracy')
plt.plot(history.history['val_loss'], label='GRU Val Loss')
plt.legend()
plt.show()
```
